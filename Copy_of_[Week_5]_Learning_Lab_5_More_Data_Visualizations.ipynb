```python
# Setup
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

plt.rcParams["figure.figsize"] = (8, 5)

# Load seaborn penguins dataset
df = sns.load_dataset("penguins")
df.head()
```

## 1) Histogram — Distribution of a Numeric Feature

**Instructions:** Plot the histogram by running the cell below.

```python
# Histogram
sns.histplot(data=df, x="flipper_length_mm", bins=20, kde=False)
plt.title("Distribution of Flipper Length (mm)")
plt.xlabel("flipper_length_mm"); plt.ylabel("count")
plt.show()
```

**Questions:**
- Where do most penguins’ flipper lengths seem to fall?
- If you were comparing two species of penguins, how might you use this type of chart to help you?
- If you wanted to describe an average penguin in this dataset, what flipper length would you mention?

## **Answers (Respond with 2-3 sentences per each question):**
- Most penguins flipper lengths seem to fall between 185mm and 205mm with a peak around 195 mm.
- If comparing two species, we could plot histograms for each species separately on the same chart or side-by-side to see the differences in their flipper length distributions and central tendencies.
- Based on the histogram, an average penguin in this dataset would have a flipper length of approximately 195mm, as this is where the highest frequency of data points occurs.

## 2) Correlation Matrix — Relationships Among Numeric Features


**Instructions:** Run the cell below to display the correlation matrix!

```python
# Correlation matrix
num = df.select_dtypes(include=[np.number]).dropna()
corr = num.corr().round(2)

sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Matrix — Penguin Measurements")
plt.show()
```

**Questions:**
- Which penguin measurements show the strongest positive correlation? How about the strongest negative correlation?
- Are any variables nearly uncorrelated? What might that imply?

## **Answers (Respond with 2-3 sentences per each question):**
- The strongest positive correlation is between flipper length and body mass (0.87). The strongest negative correlation is between bill depth and flipper length (-0.58).
- Bill length and bill depth are nearly uncorrelated (-0.24). This might imply that these two measurements are largely independent of each other for penguins in this dataset.

## 3) Line Graph — Trend Over Time

**Instructions:** Run the cell below to display the line graph!

```python
df = sns.load_dataset("penguins").dropna(subset=["flipper_length_mm", "body_mass_g", "species"])

# Sort values so the line graph connects in order
df = df.sort_values("flipper_length_mm")

plt.figure(figsize=(8,6))
sns.lineplot(
    data=df,
    x="flipper_length_mm",
    y="body_mass_g",
    hue="species",
    linewidth=2.5,
    palette="Set2"
)

plt.title("Relationship Between Flipper Length and Body Mass", fontsize=14, weight="bold")
plt.xlabel("Flipper Length (mm)", fontsize=12)
plt.ylabel("Body Mass (g)", fontsize=12)
plt.grid(alpha=0.3)
plt.legend(title="Species", loc="best")
plt.show()


```

**Questions:**
- If you saw two penguins with the same flipper length but different body masses,
what could explain the difference (think species, sex, or environment)?
- What general pattern do you notice between flipper length and body mass?
- What might explain why larger penguins also have longer flippers?

## **Answers (Respond with 2-3 sentences per each question):**
- If two penguins had the same flipper length but different body masses, this difference could be explained by their species or sex, as the line graph shows different relationships between flipper length and body mass for different species and sexes. Environmental factors could also play a role.
- The general pattern is a positive correlation: as flipper length increases, body mass tends to increase as well. This trend appears to hold true across all three species shown.
- Larger penguins likely have longer flippers because flippers are a key part of their anatomy for swimming and moving in the water. Longer flippers may be necessary to propel a larger body mass efficiently.

## 4) Pie Chart — Proportions of a Category

**Instructions:** Plot the pie chart by running the cell below.

```python
# Pie chart: species composition
counts = df["species"].value_counts(dropna=True)
plt.figure()
plt.pie(counts.values, labels=counts.index, autopct="%1.1f%%", startangle=90)
plt.title("Species Composition")
plt.axis("equal")
plt.show()
```

**Questions:**
- Which penguin species makes up the largest part of the dataset? Which species makes up the smallest part?
- Why do you think one species might appear more often in the data than others?
- If the sample were collected differently, how might these proportions change?

## **Answers (Respond with 2-3 sentences per each question):**
- Adelie penguins make up the largest part of the dataset, and Chinstrap penguins make up the smallest part.
- One species might appear more often than others due to various factors, such as the sampling location (certain islands may have a higher population of one species), the time of year the data was collected (breeding seasons could affect presence), or simply the sampling methodology used.
- If the sample were collected from different locations, times, or with different methods, the proportions of each species could change significantly to reflect the actual distribution of penguins in those areas.

## 5) Bar Chart — Comparing Group Averages

**Instructions**: Plot the bart chart by running the cell below.

```python
# Bar chart: mean flipper length by species
sns.barplot(data=df, x="species", y="flipper_length_mm", ci=95, estimator=np.mean)
plt.title("Mean Flipper Length (mm) by Species")
plt.xlabel("species"); plt.ylabel("mean flipper_length_mm")
plt.show()
```

- Gentoo penguins might have longer flippers because they are generally larger than the other two species. Longer flippers could be an adaptation related to their size and how they navigate or hunt in their environment.
